# Show Grid Lines for Raytransfer Matrix

Here we show how to visualize the grid lines for the raytransfer matrix. The grid lines are drawn in the cross-section and along the axis of the raytransfer box or cylinder.

```python
from raysect.core.math import rotate_z
from raysect.optical import Node, World

from cherab.nagdis.inversion.raytransfer import (
    create_raytransfer_box,
    create_raytransfer_cylinder,
    plot_rtb_cross_section_grid,
    plot_rtb_grid,
    plot_rtc_grid,
)
```

```python
world = World()
```

## Raytransfer Cylinder

```python
node = Node(world)
rtc = create_raytransfer_cylinder(node)

fig, axes = plot_rtc_grid(rtc, is_plot_axis=True)

print(f"The number of bins: {rtc.bins}")
print(f"Grid shape: {rtc.material.grid_shape}")
nr, nphi, nz = rtc.material.grid_shape
print(f"Total number of radial-azimuthal bins: {nr * nphi}")
```

## Raytransfer Box

**Note that we need to avoid having the center of the voxel coincide with the origin**

This is because when we set the gradient-based derivative matrices, the origin point will be singular.

```python
node = Node(world)
rtb = create_raytransfer_box(node, dz=1.0)
fig, axes = plot_rtb_grid(rtb, is_plot_axis=True)
print(f"The number of bins: {rtb.bins}")
print(f"Grid shape: {rtb.material.grid_shape}")
nx, ny, nz = rtb.material.grid_shape
print(f"Total number of xy bins: {nx * ny}")
print(f"(dx, dy, dz) = {rtb.material.grid_steps}")
```

### Rotate Grid along the Axis

Roatating a raytransfer object along the axis is useful when the tomogram keeps the same shape during the rotation.
Here we show how to visualize the grid lines for the raytransfer box when it is rotated along the axis.

```python
node.transform = rotate_z(0 * 360 // 14)
fig, ax = plot_rtb_grid(rtb, is_plot_axis=False)
node.transform = rotate_z(1 * 360 // 14)
ax = plot_rtb_cross_section_grid(rtb, ax=ax, color="0.4", zorder=-1)
node.transform = rotate_z(2 * 360 // 14)
ax = plot_rtb_cross_section_grid(rtb, ax=ax, color="0.8", zorder=-2)

ax.format(
    xreverse=False,
    xformatter=("func", lambda x, _: f"{x * 1e3:.0f}"),
    yformatter=("func", lambda x, _: f"{x * 1e3:.0f}"),
    xtickminor=True,
    ytickminor=True,
    xlabel="$x$ [mm]",
    ylabel="$y$ [mm]",
    title="Grid rotated by 0, 1, and 2 steps of 360/14 degrees",
    grid=False,
)
```
